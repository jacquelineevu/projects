**1-- Recode 2 different variables into new categories. They can both be continuous-ish or both be nominal-ish, or one of each. Tell me what you did and explain the variable(s).**



```python
from google.colab import files # mobile data usage from https://data.mendeley.com/datasets/thjkkj45fg/1
uploaded = files.upload()
```

```python
import io

df = pd.read_csv(io.BytesIO(uploaded['mobile (3).csv']))
df.head()
```

```python
df = pd.read_csv(io.BytesIO(uploaded['mobile (3).csv']))

conditions = [
    df['Age_Group'].isin(['15-18', '18 or below', '19-22']),
    df['Age_Group'].isin(['23-30', '31-40']),
    df['Age_Group'].isin(['40-50', '41-50', '51-60']),
    df['Age_Group'].isin(['50 and above', '60 or above'])
]

choices = [
    'Youth / Young Adult',
    'Adult',
    'Middle-Aged',
    'Senior'
]

df['Age_Category'] = np.select(
    conditions,
    choices,
    default='Other'
)

df[['Age_Group', 'Age_Category']].drop_duplicates().reset_index(drop=True)
```

```python
df['Age_Category'] = pd.Categorical(
    df['Age_Category'],
    categories=['Youth / Young Adult', 'Adult', 'Middle-Aged', 'Senior'],
    ordered=True
)

display(df['Age_Category'].value_counts(dropna=False))
```

```python
df['Age_Group'].value_counts(dropna=False)
```

I recoded age group which were in ranges, into categories such as youth/young adult, adult, middle-aged, and senior.

```python
conditions = [
    df['Daily_Screen_Time'] == 'Less than 2 hours',
    df['Daily_Screen_Time'] == '2-4 hours',
    df['Daily_Screen_Time'].isin(['4-6 hours', '6-8 hours']),
    df['Daily_Screen_Time'] == 'More than 8 hours'
]

choices = [
    'Very Low Usage',
    'Low Usage',
    'Moderate Usage',
    'High Usage'
]

df['Screen_Time_Category'] = np.select(
    conditions,
    choices,
    default='Other'
)

df['Screen_Time_Category'] = pd.Categorical(
    df['Screen_Time_Category'],
    categories=['Very Low Usage', 'Low Usage', 'Moderate Usage', 'High Usage'],
    ordered=True
)

df['Screen_Time_Category'].value_counts(dropna=False)
```

I recoded daily screen time into categories such as moderate, low, very low, and high usage instead of the range given.

**3-- Run a correlation of one variable with another variable; make all of the recodes necessary to make the correlation as easy to interpret as possible; and explain your results.**

```python
distraction_map = {
    'Not at all': 0,
    'Slightly': 1,
    'Moderately': 2,
    'Very': 3
}

df['Distraction_Code'] = df['Distraction_Level'].map(distraction_map)

df['Distraction_Code'].corr(df['Productivity_Rating'])
```

The correlation between distraction levels (mapped on a numeric value of 0-3) and productivity rating (mapped on a numeric scale of 1-5). The correlation is -0.41538818514381404. This means there is a negative relationship, as distraction levels increase, productivity ratings tend to decrease.
